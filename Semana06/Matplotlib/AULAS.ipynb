matplotlib ->biblioteca desenvolvidas para gerar graficos

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
t = np.linspace(0,5,10)
t
```

```python
y = np.cos(t)
```

```python
plt.plot(t,y)
plt.show()
```

```python
a = np.linspace(0,2*np.pi,100)
b = np.cos(4*a)
plt.plot(a,b)

plt.title("grafico do cosseno")
plt.xlabel('tempo')
plt.ylabel('amplitude')
plt.show()
```
