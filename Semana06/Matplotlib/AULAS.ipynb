Quanto maior for a quantidade de valores de dados que você tem no arry  mais nitisdo e fluido é o gráfico gerado.

```python
a = np.linspace(0,2*np.pi,500)
b = np.cos(4*a)
plt.plot(a,b)

plt.title("Gráfico do Cosseno")
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
```

O matplotlib funciona em camadas.Temos a camada 'Figure' que é o que a gente realmente vê, sem o gráfico.A camada 'Axes' é onde os gráficos são gerados. E tem a camada 'Axis' que é o eixo.

```python
a = np.linspace(0,2*np.pi,500)
b = np.cos(4*a)


plt.figure('COSSENO')
plt.plot(a,b)
plt.title("Gráfico do Cosseno")
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
temp = np.arange(0,100)
y = np.exp(temp)

plt.figure('Gráfico', figsize=(6,4))
plt.plot(temp, y)
```

```python
temp = np.arange(0,100)
y = np.cos(temp)
y1 = np.sin(temp)

plt.figure('Gráfico', figsize=(6,4))
plt.plot(temp, y)
plt.plot(temp,y1)
```

Vamos melhorar a visualização das duas funções no gráfico.

```python
temp = np.linspace(0,2*np.pi,100)
y = np.cos(temp)
y1 = np.sin(temp)

plt.figure('Gráfico', figsize=(6,4))
plt.plot(temp, y)
plt.plot(temp,y1)
plt.title('Gráfico do Seno e Cosseno')
plt.xlabel('Eixo do Tempo')
plt.ylabel('Eixo da Amplitude')
plt.grid()
```

```python
x = np.linspace(1, 5 ,500)
lg = np.log10(x)

plt.style.use('ggplot')

fig,axe = plt.subplots(figsize = (7,4))
axe.plot(x,lg,lw = 1.2)

axe.text(2.6,0.38,'P(2.5,0.4)',color='black')
axe.text(3,0.43,'Logarítimo $Y = log_{10}x$',fontsize=10,bbox=dict(facecolor='red',alpha=0.5),color='black')

axe.annotate('P(2.5,0.4)', xy=(2.5,0.4),fontsize = 16,xytext=(0.5,0.5),arrowprops=dict(facecolor='red'),color='black')

axe.plot([0,2.5],[0.4,0.4], color='orange',linestyle='--',lw='1.0')
axe.plot([2.5,2.5],[0,0.4], color='orange',linestyle='--',lw='1.0')

axe.plot(2.5,0.4,marker='o',color='black')

axe.set_xticks(np.arange(0,5.5,0.5))
axe.set_title("Gráfico do Logarítimico")
axe.set_xlabel('Eixo X')
axe.set_ylabel('Eixo Y')
```
