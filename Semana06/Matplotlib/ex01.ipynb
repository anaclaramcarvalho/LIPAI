 Visualização do CSV de classificação

Faça os gráficos: 

Gráfico de barras: contagem por real_class. 


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv("C:/Users/anacl/OneDrive/Documents/UFU/LIPAI/Semana06/Matplotlib/classification_results_trial_0001.csv")
```

```python
data['real_class'].value_counts().plot(kind='bar', color=['skyblue', 'salmon'])

plt.title('Contagem por Classe Real')
plt.xlabel('Classe')
plt.ylabel('Quantidade')
plt.xticks(rotation=0)

plt.show()
```

Gráfico de barras: contagem por predicted_class. 



```python
data['predicted_class'].value_counts().plot(kind='bar', color=['lightgreen', 'coral'])

plt.title('Contagem por Classe Predita')
plt.xlabel('Classe')
plt.ylabel('Quantidade')
plt.xticks(rotation=0)

plt.show()
```

Histograma de prob_benign. 

```python
data['prob_benign'].hist(bins=20, color='skyblue', edgecolor='black')

plt.title('Distribuição das Probabilidades (Benign)')
plt.xlabel('Probabilidade')
plt.ylabel('Frequência (Quantidade de Imagens)')
plt.grid(False) 

plt.show()
```

Histograma de prob_malign. 

```python
data['prob_malign'].hist(bins=20, color='salmon', edgecolor='black')

plt.title('Distribuição das Probabilidades (Malign)')
plt.xlabel('Probabilidade')
plt.ylabel('Frequência')
plt.grid(False)

plt.show()
```

Scatter plot: X=prob_benign, Y=prob_malign, com cor diferente para acerto vs erro.

```python
data['acerto'] = data['real_class'] == data['predicted_class']

# Cria o scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(data[data['acerto'] == True]['prob_benign'], 
            data[data['acerto'] == True]['prob_malign'], 
            color='green', label='Acerto', alpha=0.5)

plt.scatter(data[data['acerto'] == False]['prob_benign'], 
            data[data['acerto'] == False]['prob_malign'], 
            color='red', label='Erro', alpha=0.5)

plt.title('Relação entre Probabilidades (Benign vs Malign)')
plt.xlabel('Probabilidade Benign')
plt.ylabel('Probabilidade Malign')
plt.legend()
plt.show()
```

Qual erro é mais comum (FP ou FN)? Crie um gráfico que ajude a visualizar 
isso. 
 

```python
FP = len(data[(data['real_class'] == 'benign') & (data['predicted_class'] == 'malign')])
FN = len(data[(data['real_class'] == 'malign') & (data['predicted_class'] == 'benign')])

FP
FN
```

```python
erros_tipos = ['FP (Falso Positivo)', 'FN (Falso Negativo)']
contagem_erros = [FP, FN]

plt.figure(figsize=(6, 5))
plt.bar(erros_tipos, contagem_erros, color=['orange', 'red'])
plt.title('Comparação de Erros: FP vs FN')
plt.ylabel('Quantidade')
plt.show()
```

E em contexto médico: qual é mais preocupante e por quê?

O Falso Negativo ocorre quando o modelo classifica uma lesão maligna como benigna.O paciente deixa de ser encaminhado para uma biópsia ou tratamento imediato. Como muitas lesões malignas progridem rapidamente, esse tempo perdido pode reduzir as chances de sucesso do tratamento e a taxa de sobrevida.O vai paciente acredita estar saudável e pode negligenciar sintomas que pioram, permitindo que a doença alcance estágios graves.
