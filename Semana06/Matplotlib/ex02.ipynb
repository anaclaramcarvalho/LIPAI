Curvas de treino e validação (accuracy e loss) 

Contexto: Ao treinar modelos de Machine Learning ou Deep Learning, é 
fundamental monitorar como o modelo aprende. As métricas de loss (perda) e 
accuracy (acurácia) são registradas tanto para o conjunto de treinamento quanto 
para o conjunto de validação em cada época (um ciclo completo de treinamento 
sobre os dados). Analisar essas tendências nos ajuda a entender se o modelo está 
aprendendo bem, se está se ajustando demais aos dados de treinamento 
(overfitting) ou se não está aprendendo o suficiente (underfitting). 


Gere um gráfico semelhante ao de referência com as curvas de train_loss e val_loss 
e train_acc e val_acc. Use o eixo x como épocas. Use o arquivo de dados 
metrics.csv. 

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = pd.read_csv("C:/Users/anacl/OneDrive/Documents/UFU/LIPAI/Semana06/Matplotlib/metrics.csv")
```

```python
data
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))


ax1.plot(data['train_acc'], label='train')
ax1.plot(data['val_acc'], label='valid')
ax1.set_title('model accuracy')
ax1.set_ylabel('accuracy')
ax1.set_xlabel('epoch')
ax1.legend(loc='upper left')

ax2.plot(data['train_loss'], label='train')
ax2.plot(data['val_loss'], label='valid')
ax2.set_title('model loss')
ax2.set_ylabel('loss')
ax2.set_xlabel('epoch')
ax2.legend(loc='upper left')
plt.show()
```
