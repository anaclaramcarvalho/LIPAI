1. Quantas imagens são “benign” e “malign” em real_class?


2. Identifique em quais imagens o modelo errou a predição.

```python
falsos_positivos = data[(data['real_class'] == 'benign') & (data['predicted_class'] == 'malign')]
falsos_negativos = data[(data['real_class'] == 'malign') & (data['predicted_class'] == 'benign')]
```

3. Verifique se o modelo estava confiante mesmo quando errou. 

```python
erros = data[data['real_class'] != data['predicted_class']]
```

```python
erros
```

```python
resumo_erros = erros.groupby(['real_class', 'predicted_class']).size()
print("Distribuição dos tipos de erro:")
print(resumo_erros)
```

4. Calcule: 

a. TP (real malign, previsto malign) 

```python
tp_malign = data[(data['real_class'] == 'malign') & (data['predicted_class'] == 'malign')]
```

```python
quantidade_tp = len(tp_malign)
quantidade_tp
```

b. TN (real benign, previsto benign) 

```python
tn_benign = data[(data['real_class'] == 'benign') & (data['predicted_class'] == 'benign')]

```

```python

quantidade_tn = len(tn_benign)
quantidade_tn
```

c. FP (real benign, previsto malign)

```python
fp_benign = data[(data['real_class'] == 'benign') & (data['predicted_class'] == 'malign')]
```

```python
quantidade_fp = len(fp_benign)
quantidade_fp
```

d. FN (real malign, previsto benign) 

```python
fp_malign = data[(data['real_class'] == 'malign') & (data['predicted_class'] == 'benign')]
```

```python
quantidade_fn = len(fp_malign)
quantidade_fn
```

5. Calcule: 

a. Acurácia: (TP+TN)/(TP+TN+FP+FN) 

```python
total = quantidade_tp + quantidade_tn + quantidade_fp + quantidade_fn
acuracia = (quantidade_tp + quantidade_tn) / total if total > 0 else 0
acuracia
```

b. Precisão: TP/(TP+FP) 

```python
precisao = quantidade_tp / (quantidade_tp + quantidade_fp)
precisao
```

c. Recall: TP/(TP+FN) 


```python
recall = quantidade_tp / (quantidade_tp + quantidade_fn)
recall
```

d. Especificidade: TN/(TN+FP) 

```python
especificidade = quantidade_tn / (quantidade_tn + quantidade_fp)
especificidade
```

6. Encontre as 5 imagens benign com menor prob_benign. O que isso pode indicar? 


```python
benignos = data[data['real_class'] == 'benign']

menores_confiancas = benignos.sort_values(by='prob_benign', ascending=True).head(5)
menores_confiancas
```

O modelo comete erros de classificação apresentando alta confiança. Isso sugere que existem padrões visuais nas lesões benignas que ele está aprendendo como se fossem características de malignidade.

7. Encontre as 5 imagens malign com maior prob_benign. O que isso pode indicar?

```python
benignos = data[data['real_class'] == 'benign']

maiores_confiancas = benignos.sort_values(by='prob_benign', ascending=False).head(5)
maiores_confiancas
```

O modelo está classificando lesões que são clinicamente malignas como benignas. Este é um tipo de erro que deve ser minimizado , pois pode levar à falta de tratamento ou atraso no diagnóstico. As imagens malignas com alta prob_benign podem apresentar características visuais muito discretas, que não são facilmente capturadas pelo extrator de características do modelo.
