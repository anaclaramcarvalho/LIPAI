# ***Os Desafios em Algoritmos de ML*** 

#### ***Quantidade insuficiente de dados de treinamento***
A quantidade de dados disponíveis para treinamento é um fator fundamental para o sucesso de um modelo de aprendizado de máquina. Quando há poucos exemplos para treinar, o algoritmo pode ter dificuldades para identificar padrões representativos, o que compromete sua capacidade de fazer boas previsões em novos dados.

Esse problema é especialmente relevante em contextos nos quais a coleta de dados é cara, demorada ou limitada, como em aplicações médicas, industriais ou 
científicas.

#### ***Principais problemas causados pela falta de dados*** 
Overfitting: 
- O modelo se ajusta excessivamente aos dados de treinamento e passa a memorizar exemplos específicos, perdendo capacidade de generalização. Nesse caso, ele apresenta bom desempenho nos dados usados no treinamento, mas falha ao lidar com novos exemplos. 

Underfitting: 
- O modelo é simples demais ou não consegue aprender adequadamente a relação entre os dados de entrada e a saída. Isso resulta em baixo desempenho tanto nos dados de treinamento quanto nos dados de teste. 

Baixa performance: 
- Com poucos dados, o modelo pode não conseguir capturar padrões importantes, o que reduz sua precisão e confiabilidade. 

Viés nos resultados: 
- Amostras pequenas podem não representar corretamente a população real, fazendo com que o modelo aprenda padrões distorcidos ou incompletos.

#### ***Estratégias para reduzir os impactos da escassez de dados***
Data Augmentation: 
- Consiste em criar variações sintéticas dos dados já existentes. Em imagens, por exemplo, isso pode incluir rotações, espelhamentos, cortes ou alterações de brilho, aumentando artificialmente o conjunto de treinamento.

Transfer Learning: 
- Consiste em utilizar modelos previamente treinados em grandes bases de dados e adaptá-los para um problema específico. Essa abordagem é muito útil quando há poucos dados disponíveis. 

Técnicas de regularização: 
- São métodos aplicados para reduzir overfitting e melhorar a generalização do modelo, mesmo em cenários com dados limitados.
