# ***Algoritmos de Classificação Supervisionados***

São técnicas de aprendizado de máquina em que o algoritmo aprende a partir de dados rotulados para fazer previsões. Isso significa que, durante o treinamento, o modelo recebe exemplos de entrada junto com suas respectivas depois serão usados para prever novos casos. 

Esse processo permite que o algoritmo associe características dos dados de 
entrada aos rótulos fornecidos, criando um modelo capaz de generalizar para 
dados nunca vistos. A eficácia desse tipo de aprendizado depende fortemente da 
qualidade e da quantidade dos dados de treinamento disponíveis.

#### ***Principais características do aprendizado supervisionado***
Conjunto de treinamento: 
- Dados com rótulos conhecidos, utilizados para treinar o modelo.

Processo de aprendizagem: 
- O algoritmo aprende padrões nos dados de entrada, associa esses padrões aos 
rótulos fornecidos e cria um modelo capaz de fazer previsões em novos dados. 

Tipos de problemas: 
- Classificação: previsão de categorias discretas 

- Regressão: previsão de valores contínuos 

Principais algoritmos: 
- Árvores de Decisão 
- Random Forest 
- Support Vector Machines (SVM) 
- Redes Neurais Artificiais 
- k-Nearest Neighbors (k-NN)

#### ***Regressão Linear*** 
É um dos algoritmos mais básicos e amplamente utilizados em 
aprendizado supervisionado, principalmente para problemas de regressão. 

Objetivo: 
- Encontrar uma relação linear entre variáveis independentes (features) e uma 
variável dependente (alvo). 

Tipos: 
- Regressão Linear Simples: uma única variável independente 
- Regressão Linear Múltipla: múltiplas variáveis independentes 

Funcionamento: 
- A regressão linear ajusta uma linha reta aos dados minimizando o erro quadrático médio. Para isso, calcula coeficientes para cada variável independente e utiliza equações como: 
- Regressão simples: y = mx + b 
- Regressão múltipla: y = b₀ + b₁x₁ + b₂x₂ + ... + b x 

Vantagens: 
- Simples de implementar e interpretar 
- Computacionalmente eficiente 
- Boa base para entender conceitos mais complexos 

Limitações: 
- Assume relação linear entre variáveis 
- Sensível a outliers 
- Pode ter desempenho inferior em relações não lineares


#### ***Regressão Polinomial***

É uma extensão da regressão linear que permite modelar relações não lineares entre variáveis.

***Características principais*** 

Objetivo: 
- Ajustar uma curva polinomial aos dados para capturar relações não lineares. 

Funcionamento: 
- Esse método adiciona termos polinomiais às variáveis independentes, como x², x³ e outros graus,utilizando equações do tipo: 
         y = b₀ + b₁x + b₂x² + b₃x³ + ... + b xⁿ 

Embora o modelo represente relações curvas, o ajuste ainda é feito com base no método dos mínimos quadrados. 

Vantagens: 
- Maior flexibilidade que a regressão linear 
- Capaz de modelar curvas e padrões mais complexos 
- Mantém certa interpretabilidade do modelo 

Limitações: 
- Risco de overfitting com graus muito altos 
- Requer mais dados para treinar de forma eficaz 
- Sensível a outliers

#### ***Regressão Logística*** 
Apesar do nome, a regressão logística é utilizada principalmente para problemas de classificação binária. 

***Características principais***

Objetivo: 
- Estimar a probabilidade de um exemplo pertencer a uma determinada classe. 

Funcionamento: 
- A regressão logística utiliza a função sigmóide para transformar valores contínuos em probabilidades entre 0 e 1. Em seguida, aplica-se um limite de decisão, geralmente 0,5, para determinar a classe final. O treinamento costuma ser feito com base no método de máxima verossimilhança. 

Vantagens: 
- Simples de implementar e interpretar 
- Fornece probabilidades como saída 
- Computacionalmente eficiente 
- Menos suscetível a overfitting em comparação com modelos mais complexos 

Limitações: 
- Assume uma fronteira de decisão linear 
- Pode ter desempenho inferior em relações altamente não lineares 
- Requer tratamento adequado de variáveis categóricas 

#### ***Árvores de Decisão*** 
As árvores de decisão são algoritmos de aprendizado de máquina usados para predição, mapeando observações sobre um item para conclusões sobre seu valor alvo. Seu nome vem da estrutura em forma de árvore. 

Estrutura básica 
- Nó raiz e nós internos: representam testes em atributos 
- Ramos: representam os resultados dos testes 
- Folhas: representam classes, em problemas de classificação, ou valores, em problemas de regressão 

Funcionamento 
- Esse algoritmo realiza um particionamento recursivo dos dados, dividindo-os repetidamente com base em regras de decisão. 

Critérios de divisão: 
- Para classificação: Gini Impurity ou Entropia/Ganho de Informação 
- Para regressão: Redução de variância ou erro quadrático médio (MSE) 

Vantagens 
- Fácil interpretação e visualização 
- Exige pouco pré-processamento 
- Lida bem com dados numéricos e categóricos 
- Baixo custo computacional para previsões 

Desvantagens 
- Tendência ao overfitting, especialmente em árvores muito profundas 
- Pequenas variações nos dados podem gerar árvores bastante diferentes 
- Pode apresentar viés para classes dominantes em problemas desbalanceados

#### ***Por que estudar esses algoritmos?*** 
É fundamental porque eles representam a base de muitos sistemas de aprendizado de máquina utilizados em aplicações reais. Além disso, compreender seus princípios ajuda no desenvolvimento de soluções mais 
robustas e na escolha adequada de modelos para diferentes tipos de problemas. 

Esses métodos são amplamente utilizados em áreas como diagnóstico médico, reconhecimento de padrões, análise financeira, classificação de documentos, 
sistemas de recomendação e previsão de variáveis contínuas.
