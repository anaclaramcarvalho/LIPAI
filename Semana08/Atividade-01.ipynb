***APRENDIZADO SUPERVISIONADO***

Técnicas de aprendizado de máquina em que o algoritmo aprende a partir de dados rotulados para fazer previsões.

Ou seja, durante o treinamento o modelo recebe exemplos de entrada junto com suas respectivas saídas corretas,aprendendo padrões que depois serão usados para prever novos casos.

Esse processo permite que o algoritmo associe características dos dados de entrada aos rótulos fornecidos, criando um modelo capaz de generalizar para dados nunca vistos. A eficácia desse tipo de aprendizado depende fortemente da qualidade e da quantidade dos dados de treinamento disponíveis.

***REGRESSÃO LINEAR***

A regressão linear é um dos algoritmos mais básicos e amplamente utilizados em 
aprendizado supervisionado, principalmente para problemas de regressão.

OBJETIVO: 
Encontrar uma relação linear entre variáveis independentes (features) e uma variável dependente (alvo). 

TIPOS: 

● Regressão Linear Simples: uma única variável independente 

● Regressão Linear Múltipla: múltiplas variáveis independentes 

FUNCIONAMENTO:

A regressão linear ajusta uma linha reta aos dados minimizando o erro quadrático médio. Para isso, calcula coeficientes para cada variável independente e utiliza equações como: 

● Regressão simples

● Regressão múltipla

VANTAGENS:

● Simples de implementar e interpretar 

● Computacionalmente eficiente 

● Boa base para entender conceitos mais complexos 

LIMITAÇÕES: 

● Assume relação linear entre variáveis 

● Sensível a outliers
 
● Pode ter desempenho inferior em relações não lineares 

***REGRESSÃO POLINOMIAL***

A regressão polinomial é uma extensão da regressão linear que permite modelar relações não lineares entre variáveis.

OBJETIVO: 

Ajustar uma curva polinomial aos dados para capturar relações não lineares. 

FUNCIONAMENTO:

Esse método adiciona termos polinomiais às variáveis independentes, como x², x³ e outros graus, utilizando equações.
Embora o modelo represente relações curvas, o ajuste ainda é feito com base no método dos mínimos quadrados. 

VANTAGENS:

● Maior flexibilidade que a regressão linear 

● Capaz de modelar curvas e padrões mais complexos 

● Mantém certa interpretabilidade do modelo 

LIMITAÇÕES:

● Risco de overfitting com graus muito altos

● Requer mais dados para treinar de forma eficaz 

● Sensível a outliers 

***REGRESSÃO LOGÍSTICA***

Apesar do nome, a regressão logística é utilizada principalmente para problemas de classificação binária. 

OBJETIVO:

Estimar a probabilidade de um exemplo pertencer a uma determinada classe. 

FUNCIONAMENTO:

A regressão logística utiliza a função sigmóide para transformar valores contínuos em probabilidades entre 0 e 1. Em seguida, aplica-se um limite de decisão, 
geralmente 0,5, para determinar a classe final. O treinamento costuma ser feito 
com base no método de máxima verossimilhança. 

VANTAGENS:

● Simples de implementar e interpretar

● Fornece probabilidades como saída

● Computacionalmente eficiente

● Menos suscetível a overfitting em comparação com modelos mais complexos

LIMITAÇÕES:

● Assume uma fronteira de decisão linear

● Pode ter desempenho inferior em relações altamente não lineares

● Requer tratamento adequado de variáveis categóricas

***ÁRVORES DE DECISÃO***

As árvores de decisão são algoritmos de aprendizado de máquina usados para 
predição, mapeando observações sobre um item para conclusões sobre seu valor alvo. Seu nome vem da estrutura em forma de árvore. 

ESTRUTURA BÁSICAS

● Nó raiz e nós internos: representam testes em atributos 

● Ramos: representam os resultados dos testes

● Folhas: representam classes, em problemas de classificação, ou valores, em problemas de regressão 

FUNCIONAMENTO

Esse algoritmo realiza um particionamento recursivo dos dados, dividindo-os 
repetidamente com base em regras de decisão. 

CRITÉRIOS DE DIVISÃO

● Para classificação: Gini Impurity ou Entropia/Ganho de Informação

● Para regressão: Redução de variância ou erro quadrático médio (MSE) 

VANTAGENS

● Fácil interpretação e visualização 

● Exige pouco pré-processamento 

● Lida bem com dados numéricos e categóricos 

● Baixo custo computacional para previsões 

DESVANTAGENS

● Tendência ao overfitting,especialmente em árvores muito profundas

● Pequenas variações nos dados podem gerar árvores bastante diferentes 

● Pode apresentar viés para classes dominantes em problemas desbalanceados

***REFLEXÃO***

Após realizar os experimentos, reflita sobre as seguintes questões: 

● Como a escolha do algoritmo influencia o desempenho final? 

Determina a capacidade do modelo de capturar padrões,ou seja,complexidade e a velocidade de processamento. Um algoritmo inadequado pode causar underfitting, sendo simples demais para os dados ou decora o ruído dos dados.

● Quais métricas são mais adequadas para comparar classificadores? 

Depende do equilíbrio dos dados. As principais são Acurácia, Precisão e Recall  e a Matriz de Confusão.

● Como a qualidade dos dados pode impactar os resultados? 

Dados ruins geram modelos ruins . Ruídos, valores ausentes ou desbalanceamento levam a previsões enviesadas, baixa confiabilidade e falhas na generalização para situações reais.

● Em quais situações modelos mais simples podem ser mais vantajosos que modelos mais complexos?

Modelos simples são vantajosos quando há poucos dados, necessidade de baixa latência , ou quando a interpretabilidade é prioritária. Eles também são menos propensos ao overfitting.
