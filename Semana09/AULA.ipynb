# ***Redes Neurais Artificiais e PyTorch***

As redes neurais artificiais (RNAs) são modelos computacionais inspirados no funcionamento do cérebro humano, capazes de aprender padrões a partir de 
dados. Esses modelos são formados por unidades de processamento chamadas neurônios artificiais, que recebem entradas, aplicam operações matemáticas e 
produzem saídas. 

***Composta por***

Camada de entrada: 
- Recebe os dados brutos do problema. 

Camadas ocultas: 
- Realizam o processamento intermediário das informações. 

Camada de saída: 
- Produz o resultado final, como uma classificação ou uma predição. 

***Além disso,possuem:*** 

Pesos e viés (bias): 
- Parâmetros ajustados durante o treinamento para minimizar os erros do modelo. 

Funções de ativação: 
- Introduzem não linearidade no modelo, permitindo aprender relações mais complexas nos dados. Exemplos: ReLU, Sigmoid e Tanh. 

O desempenho de uma rede neural depende da arquitetura escolhida, da qualidade dos dados, da função de perda, do otimizador e de outros fatores relacionados ao treinamento.

#### ***Principais características das Redes Neurais Artificiais*** 
Aprendizado a partir de dados: 
- A rede aprende padrões automaticamente com base em exemplos fornecidos durante o treinamento. 

Arquitetura em camadas: 
- Os dados passam por diferentes camadas de processamento até chegar ao resultado final. 

Ajuste de parâmetros: 
- Os pesos e vieses são atualizados para reduzir o erro do modelo. 

Capacidade de modelar relações complexas: 
- Graças às funções de ativação e às múltiplas camadas, conseguem representar padrões não lineares. 

Aplicações diversas: 
- Podem ser utilizadas em tarefas de classificação, regressão, visão computacional, processamento de linguagem natural, séries temporais e muito mais.

#### ***Tipos principais*** 
Perceptron Multicamadas (MLP): 
- Rede do tipo feedforward com uma ou mais camadas ocultas. 
- É utilizada em tarefas como classificação e regressão. 

Redes Neurais Convolucionais (CNNs): 
- Especializadas em processar dados organizados em grade, como imagens. 
- Utilizam filtros (kernels) para detectar características locais. 

Redes Neurais Recorrentes (RNNs): 
- Indicadas para lidar com dados sequenciais, como textos e séries temporais. 
- Variantes como LSTM e GRU ajudam a reduzir problemas como vanishing gradient. 

Transformers: 
- Arquitetura utilizada em modelos modernos como GPT e BERT. 
- Baseiam-se em mecanismos de autoatenção para processar dependências de longo 
alcance. 

#### ***Treinamento de uma Rede Neural*** 
Backpropagation: 
- Algoritmo utilizado para ajustar os pesos da rede com base no gradiente do erro. 

Função de perda (loss): 
- Mede a diferença entre as previsões do modelo e os valores reais. 

Exemplos: 
MSE para regressão 
Cross-Entropy para classificação 

Otimizadores: 
- Responsáveis por atualizar os pesos com o objetivo de minimizar a função de 
perda. 

Exemplos: 
SGD 
Adam 

Regularização: 
- Técnicas como Dropout e penalização L2 ajudam a evitar overfitting. 

#### ***Por que estudar Redes Neurais Artificiais?*** 
O estudo das redes neurais artificiais é fundamental porque elas estão na base de muitos sistemas modernos de inteligência artificial. Esses modelos são capazes de resolver problemas complexos em diferentes áreas, como: 
- Diagnóstico médico 
- Reconhecimento de padrões 
- Processamento de linguagem natural 
- Visão computacional 
- Sistemas de recomendação 
- Previsão de séries temporais

Além disso, compreender os conceitos fundamentais de redes neurais ajuda no entendimento de arquiteturas mais avançadas e na escolha adequada de soluções 
para diferentes problemas.

***PyTorch*** é uma biblioteca amplamente utilizada para desenvolvimento de modelos de deep learning. Ela oferece ferramentas para definição de redes neurais, cálculo automático de gradientes, treinamento de modelos e manipulação de tensores. 

Na construção de modelos em PyTorch, normalmente criamos classes que herdam de nn.Module, definimos as camadas no método __init__ e o fluxo de dados no 
método forward.
