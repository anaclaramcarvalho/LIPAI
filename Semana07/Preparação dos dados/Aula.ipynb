# ***Preparação dos Dados***

A qualidade de um modelo de IA depende diretamente da forma como os dados são tratados: entender se um dado é discreto ou contínuo, como normalizar escalas
e como dividir conjuntos de dados para evitar o overfitting (quando o modelo apenas "decora" os exemplos).

### Tipos de Dados 

***Dados Discretos*** 

São valores que podem ser contados e representam quantidades inteiras. 

Características principais: 
- Assumem valores finitos ou contáveis 
- Não podem ser divididos em partes menores mantendo o significado 
- Exemplos: número de filhos, quantidade de produtos, número de 
ocorrências 

***Dados Contínuos*** 

São valores que podem assumir qualquer número dentro de um intervalo. 

Características principais: 
- Podem assumir infinitos valores em um intervalo 
- Podem ser divididos em partes menores mantendo o significado 
- Exemplos: altura, peso, temperatura, tempo 

É importante notar que a distinção entre discreto e contínuo pode afetar a escolha de métodos estatísticos e visualizações apropriadas para análise.

Em problemas de classificação podem ser categorizados da seguinte forma: 

***Numéricos:*** 

- Contínuos: valores reais como altura, peso, temperatura 

- Discretos: valores inteiros como contagens, quantidade de ocorrências 

***Categóricos:***

- Nominais: categorias sem ordem natural (cores, tipos de sangue) 

- Ordinais: categorias com ordem definida (pequeno, médio, grande) 

***Temporais:*** 

- Datas e horários 

- Séries temporais 

***Textuais:*** 

- Strings não estruturadas 

- Documentos de texto 

***Binários:*** 

- Valores booleanos (verdadeiro/falso) 

- Indicadores de presença/ausência 

A compreensão adequada dos tipos de dados é fundamental para escolher as 
técnicas apropriadas de pré-processamento e os algoritmos de classificação. 

### Transformação de Dados

A transformação de dados é uma etapa crucial no pré-processamento, garantindo 
que os dados estejam em formato adequado para análise. 

***Normalização:*** Ajusta os valores para uma escala comum, geralmente entre 0 e 1 

***Padronização:*** Transforma os dados para terem média zero e desvio padrão 
unitário 

***Codificação:*** 

- One-Hot Encoding: Converte variáveis categóricas em colunas binárias 

- Label Encoding: Atribui números para categorias ordinais 

***Tratamento de Valores Ausentes:*** 

- Imputação pela média, mediana ou moda 

- Remoção de registros incompletos 

***Redução de Dimensionalidade:*** 

- PCA (Análise de Componentes Principais) 

- Seleção de características relevantes 

***Balanceamento de Classes:*** 

- Oversampling 

- Undersampling 

- Técnicas híbridas (SMOTE)

Cada transformação deve ser escolhida considerando a natureza dos dados e os 
requisitos do algoritmo de aprendizado. 

### Normalização de Dados 

A normalização é um processo essencial no pré-processamento de dados que 
ajusta os valores para uma escala comum, facilitando a comparação e análise.

***Min-Max Scaling***

Transforma os dados para um intervalo específico (geralmente [0,1]): 

- x_norm = (x - x_min) / (x_max - x_min) 

***Z-Score (Standardization)***

Padroniza os dados para terem média zero e desvio padrão unitário: 

- x_std = (x - média) / desvio_padrão 

***Normalização por Decimal Scaling***

Move o ponto decimal dos valores para obter números menores que 1: 

- x_norm = x / 10^j 

onde j é o menor número inteiro tal que max(|x_norm|) < 1 

***Benefícios da normalização:***

- Melhora a convergência de algoritmos de aprendizado de máquina 

- Evita que atributos com valores maiores dominem o aprendizado 

- Facilita a comparação entre diferentes características 

- Reduz a sensibilidade a outliers em alguns casos 

***É importante escolher o método de normalização adequado considerando:***

- A distribuição dos dados 

- A presença de outliers 

- Os requisitos do algoritmo de aprendizado 

- O domínio do problema

### ***Por que dividimos os dados em conjuntos de teste de treinamento e validação?***

Dividir os dados em conjuntos de treinamento (training) e teste (test) é uma prática importante para os algoritmos de ML com objetivo de para garantir que o modelo generalize bem para dados não vistos. 

***Avaliar a Generalização do Modelo:***
O objetivo principal de um modelo de ML é desempenhar bem em dados novos, não apenas nos dados usados para treiná-lo. Se usar todos os dados para treinar, não teríamos como medir se o modelo aprendeu padrões verdadeiros ou apenas decorou os dados (overfitting). O conjunto de teste simula dados nunca vistos, 
permitindo avaliar se o modelo é útil no mundo real. 

***Ajustar Hiperparâmetros e Selecionar Modelos:***
Usamos o conjunto de treino para treinar o modelo e o de teste para validar sua eficácia. Isso ajuda a escolher entre diferentes algoritmos e ajustar hiperparâmetros . 

***Evitar Vazamento de Dados (Data Leakage):***
Se o modelo "vê" os dados de teste durante o treinamento, ele pode se ajustar indevidamente a eles, inflando artificialmente seu desempenho. Manter um conjunto de teste separado garante uma avaliação justa.

- Divisão Clássica (Train/Test Split): Geralmente, os dados são divididos em: 
 
        ○ 70-80% para treino (ajustar o modelo). 

        ○ 20-30% para teste (avaliar desempenho). 

Em problemas com pouco dados, técnicas como ***validação cruzada 
(cross-validation)*** são mais eficientes.
