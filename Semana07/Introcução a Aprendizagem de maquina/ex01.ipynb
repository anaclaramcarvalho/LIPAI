In detail, describe the following terminologies:

a.Generative AI

É um subconjunto do aprendizado de máquina que se concentra na criação de conteúdos novos e originais, em vez de apenas analisar dados existentes. Seus principais motores são:

Transformers: Arquiteturas que revolucionaram o processamento de linguagem natural.

GANs: Onde duas redes competem  para melhorar a qualidade do conteúdo.

Modelos de Difusão: Que criam dados novos revertendo um processo de adição de ruído, sendo muito eficazes para imagens de alta resolução.

b.NLP
É um campo especializado dentro do aprendizado de máquina focado na interação entre a linguagem humana e os computadores. O NLP fornece ferramentas para extrair insights e significado de dados não estruturados, como textos e falas
